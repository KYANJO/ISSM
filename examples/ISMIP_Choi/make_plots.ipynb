```python
# 1-1 Plot domain
# md=loadmodel('./Models_' + modelname + '/MISMIP_' + modelname + '_Parameterization2.nc')
# md = loadmodel('./Models/MISMIP_Parameterization2.nc')
md = loadmodel('ISMIP.Parameterization.nc')

# Create figure and set properties
fig = plt.figure(figsize=(20, 12)) 
fig.set_facecolor('w')
ax = fig.add_subplot(111, projection='3d')

# Define grid for interpolation
X, Y = np.meshgrid(np.arange(1, 700000, 1000), np.arange(1, 80000, 500))

# Interpolate bed, base, and surface onto the grid
bed_grid = griddata((md.mesh.x, md.mesh.y), md.geometry.bed, (X, Y), method='linear', fill_value=np.nan)
base_grid = griddata((md.mesh.x, md.mesh.y), md.geometry.base, (X, Y), method='linear', fill_value=np.nan)
surface_grid = griddata((md.mesh.x, md.mesh.y), md.geometry.surface, (X, Y), method='linear', fill_value=np.nan)

# Plot the surfaces
ax.plot_surface(X/1000, Y/1000, bed_grid, cmap='jet', edgecolor='none', linewidth=0, vmin=-2000, vmax=2500)
ax.plot_surface(X/1000, Y/1000, base_grid, cmap='jet', edgecolor='none', linewidth=0, vmin=-2000, vmax=2500)
surf = ax.plot_surface(X/1000, Y/1000, surface_grid, cmap='jet', edgecolor='none', linewidth=0, alpha=0.6, vmin=-2000, vmax=2500)

# Add colorbar
cbar = fig.colorbar(surf, ax=ax, shrink=0.4, aspect=20)
cbar.set_label('Elevation (m)', fontsize=14)

# Set the view angle
ax.view_init(30, -60)

# Set aspect ratio
ax.set_box_aspect([7, .8, 2])  # Data aspect ratio

# Lighting properties
ax.xaxis.pane.fill = False
ax.yaxis.pane.fill = False
ax.zaxis.pane.fill = False
ax.xaxis.pane.set_edgecolor('w')
ax.yaxis.pane.set_edgecolor('w')
ax.zaxis.pane.set_edgecolor('w')

# Ticks
ax.set_yticks([0, 40, 80])

# Labels
ax.set_xlabel('x (km)')
ax.zaxis.set_rotate_label(False)
ax.set_zlabel('z (m)', rotation=90)
ax.set_ylabel('y (km)')

ax.xaxis.labelpad = 50
ax.zaxis._axinfo['juggled'] = (1,2,0)
ax.zaxis.labelpad = 10

# Set font size of the axes
for item in ([ax.title, ax.xaxis.label, ax.yaxis.label, ax.zaxis.label] +
             ax.get_xticklabels() + ax.get_yticklabels() + ax.get_zticklabels()):
    item.set_fontsize(14)
    
#plt.savefig("./Graphics/plot_domain.pdf", format="pdf", bbox_inches="tight")
#plt.show()
```
